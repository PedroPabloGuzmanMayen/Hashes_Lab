```python
from generate_hashes import *
import requests
```

```python
passwds = ["admin", "123456", "hospital", "medisoft2024", "password"]
```

```python
def check_pwned(password: str):
    _, hash_full = build_hash(0, password.encode())

    hash_full = hash_full.upper()

    prefix = hash_full[:5]
    suffix = hash_full[5:]

    url = f"https://api.pwnedpasswords.com/range/{prefix}"

    response = requests.get(url)

    if response.status_code != 200:
        raise RuntimeError("Error consultando HIBP API")

    hashes = response.text.splitlines()

    for line in hashes:
        hash_suffix, count = line.split(":")
        if hash_suffix == suffix:
            return int(count)

    return 0
```

```python
print(f"{'Password':<15} {'SHA1':<64} {'Pwned Count'}")
print("-" * 100)

for pwd in passwds:
    _, h = build_hash(0, pwd.encode())
    count = check_pwned(pwd)

    print(f"{pwd:<15} {h:<64} {count}")
```
