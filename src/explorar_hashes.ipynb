# Comparación de hashes

```python
from generate_hashes import *
import pandas as pd
```

```python
algoritms = {
    'MD5': 3,
    'SHA1': 0,
    'SHA256': 1,
    'SHA3_256': 2
}

s1 = b'MediSoft-v2.1.0'
s2 = b'medisoft-v2.1.0'

rows = []

for name, mode in algoritms.items():

    d1, h1 = build_hash(mode, s1)
    d2, h2 = build_hash(mode, s2)

    rows.append({
        'Algoritmo': name,
        'Bits': len(d1) * 8,
        'HexLen': len(h1),
        'Mensaje': s1.decode(),
        'Hash': h1
    })

    rows.append({
        'Algoritmo': name,
        'Bits': len(d2) * 8,
        'HexLen': len(h2),
        'Mensaje': s2.decode(),
        'Hash': h2
    })

    if name == 'SHA256':
        bits_changed = xor_diff_bits(d1, d2)

df = pd.DataFrame(rows)
```
